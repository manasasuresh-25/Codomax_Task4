```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("cleaned data.csv")
```

```python
plt.figure(figsize=(6,5))
sns.countplot(data=df, x="type", palette="Set2")
plt.title("Distribution of Netflix Content")
plt.xlabel("Content Type")
plt.ylabel("Count")
plt.show()
```

```python
yearly = df["release_year"].value_counts().sort_index()

plt.figure(figsize=(10,5))
plt.plot(yearly.index, yearly.values, marker='o')
plt.title("Netflix Content Released by Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(6,6))
df["type"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Movies vs TV Shows")
plt.ylabel("")
plt.show()
```

```python
df["type_num"] = df["type"].map({"Movie":0, "TV Show":1})

plt.figure(figsize=(8,5))
plt.scatter(df["release_year"], df["type_num"], alpha=0.5)
plt.title("Release Year vs Content Type")
plt.xlabel("Release Year")
plt.ylabel("Content Type")
plt.yticks([0,1],["Movie","TV Show"])
plt.show()
```

```python
plt.figure(figsize=(10,6))
df["country"].value_counts().head(10).plot(kind="bar")
plt.title("Top 10 Countries")
plt.xlabel("Country")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(12,6))
df["listed_in"].value_counts().head(10).plot(kind="bar")
plt.title("Top 10 Genres")
plt.xlabel("Genre")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```
